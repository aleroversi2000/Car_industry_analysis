```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Car Sales.xlsx - car_data.csv")
df.head()
```

```python
df.shape
```

```python
# Checking duplicates presence
df.duplicated().sum()
```

```python
# Checking nulls 
df.isnull().sum()
```

```python
# Filling null values in Customer Name column 
df['Customer Name'] = df['Customer Name'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
# Checking the market geographic area
df['Dealer_Region'].unique()
```

```python
# Summary statistics 
pd.set_option('display.float_format', '{:,.2f}'.format)
df.describe()
```

```python
# Display distribution of Annual Income
sns.histplot(data=df, x="Annual Income", kde=True, bins='auto')
plt.show
```

```python
# Display distribution of Price
sns.histplot(data=df, x="Price ($)", kde=True, bins='auto')
plt.show
```

```python
# Display relationship between Price and Annual Income 
import matplotlib.ticker as mticker

ax = sns.scatterplot(data=df, x='Price ($)', y='Annual Income',
                     alpha=0.15, s=10, edgecolor=None)
ax.set_yscale('log')
ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'{x/1000:.0f}k'))
plt.show()
```
